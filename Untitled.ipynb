```python
required_skills = [
    "Python",
    "SQL",
    "Machine Learning",
    "Pandas",
    "NumPy",
    "Flask"
]

print("Required Skills:")
for skill in required_skills:
    print(skill)
```

```python
resume = """
My name is Rahul Sharma.
I am a Computer Science graduate.

I have experience in Python programming, SQL databases,
Machine Learning and Pandas. I have also worked on
data analysis projects using NumPy.

I have basic knowledge of HTML and CSS.
"""

print(resume)
```

```python
resume_text = resume.lower()

print(resume_text)
```

```python
found_skills = []

for skill in required_skills:
    if skill.lower() in resume_text:
        found_skills.append(skill)

print("Skills found in resume:")
print(found_skills)
```

```python
candidate_vector = np.array([
    1 if skill in found_skills else 0
    for skill in required_skills
])

print("Required skills:")
print(required_skills)

print("Candidate vector:")
print(candidate_vector)
```

```python
matched_skills = np.sum(candidate_vector)
total_skills = len(required_skills)

score = (matched_skills / total_skills) * 100

print("Matched Skills:", matched_skills)
print("Total Required Skills:", total_skills)
print("Match Score:", score, "%")
```

```python
threshold = 60

if score >= threshold:
    print("Result: SHORTLISTED")
else:
    print("Result: NOT SHORTLISTED")
```

```python
candidates = {
    "Rahul": """
    Python SQL Machine Learning Pandas NumPy
    """,

    "Aman": """
    Python SQL HTML CSS JavaScript
    """,

    "Priya": """
    Python Machine Learning Pandas NumPy Flask SQL
    """
}

print("Candidates:")
for name in candidates:
    print(name)
```

```python
candidate_scores = {}

for name, resume in candidates.items():

    resume_text = resume.lower()

    found_skills = []

    for skill in required_skills:
        if skill.lower() in resume_text:
            found_skills.append(skill)

    candidate_vector = np.array([
        1 if skill in found_skills else 0
        for skill in required_skills
    ])

    matched_skills = np.sum(candidate_vector)
    score = (matched_skills / len(required_skills)) * 100

    candidate_scores[name] = score

print(candidate_scores)
```

```python
ranked_candidates = sorted(
    candidate_scores.items(),
    key=lambda x: x[1],
    reverse=True
)

print("Candidate Ranking:")
print("------------------")

for rank, (name, score) in enumerate(ranked_candidates, start=1):
    print(f"{rank}. {name} - {score:.2f}%")
```

```python
print("AI RESUME SCREENING REPORT")
print("=" * 60)

threshold = 60

for rank, (name, score) in enumerate(ranked_candidates, start=1):

    if score >= threshold:
        status = "SHORTLISTED"
    else:
        status = "NOT SHORTLISTED"

    print(
        f"Rank: {rank} | "
        f"Candidate: {name} | "
        f"Score: {score:.2f}% | "
        f"Status: {status}"
    )
```

```python

job_description = """
We are looking for a Python developer with skills in
Python, SQL, Machine Learning, Pandas, NumPy and Flask.
"""

required_skills = [
    "Python",
    "SQL",
    "Machine Learning",
    "Pandas",
    "NumPy",
    "Flask"
]

print("Job requirements loaded!")
print("Required skills:", required_skills)
```

```python

import os
import numpy as np
import pandas as pd
from PyPDF2 import PdfReader

resume_folder = "resumes"
results = []

pdf_files = sorted([
    f for f in os.listdir(resume_folder)
    if f.lower().endswith(".pdf")
])

for filename in pdf_files:
    pdf_path = os.path.join(resume_folder, filename)

    try:
        reader = PdfReader(pdf_path)
        resume_text = ""

        for page in reader.pages:
            text = page.extract_text()
            if text:
                resume_text += text + "\n"

        resume_text_lower = resume_text.lower()

        candidate_vector = np.array([
            1 if skill.lower() in resume_text_lower else 0
            for skill in required_skills
        ])

        matched_skills = int(np.sum(candidate_vector))
        score = (matched_skills / len(required_skills)) * 100

        results.append({
            "Resume": filename,
            "Matched Skills": matched_skills,
            "Total Skills": len(required_skills),
            "Score (%)": round(score, 2),
            "Matched Skill Names": ", ".join(
                skill for skill, found in
                zip(required_skills, candidate_vector) if found
            )
        })

    except Exception as error:
        print(f"Could not scan {filename}: {error}")

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Score (%)",
    ascending=False
).reset_index(drop=True)

results_df.index = results_df.index + 1
results_df.index.name = "Rank"

print("Total resumes scanned:", len(results_df))
display(results_df)
```

```python

threshold = 60

shortlisted = results_df[
    results_df["Score (%)"] >= threshold
]

print("Shortlisting threshold:", threshold, "%")
print("Candidates shortlisted:", len(shortlisted))

display(shortlisted)
```

```python

import streamlit as st
import numpy as np
from PyPDF2 import PdfReader

st.set_page_config(page_title="AI Resume Screening", page_icon="📄")

st.title("AI-Based Resume Screening System")
st.write("Upload a resume to check its skill-match score.")

required_skills = [
    "Python",
    "SQL",
    "Machine Learning",
    "Pandas",
    "NumPy",
    "Flask"
]

uploaded_file = st.file_uploader("Upload a resume PDF", type=["pdf"])

if uploaded_file is not None:
    reader = PdfReader(uploaded_file)
    resume_text = ""

    for page in reader.pages:
        text = page.extract_text()
        if text:
            resume_text += text + "\n"

    resume_text_lower = resume_text.lower()

    matched_skills = [
        skill for skill in required_skills
        if skill.lower() in resume_text_lower
    ]

    score = (len(matched_skills) / len(required_skills)) * 100

    st.subheader("Screening Results")
    st.metric("Skill Match Score", f"{score:.2f}%")
    st.write("**Matched skills:**")

    if matched_skills:
        st.write(", ".join(matched_skills))
    else:
        st.write("No required skills found.")

    st.write("**Skills not found:**")
    missing_skills = [
        skill for skill in required_skills
        if skill not in matched_skills
    ]
    st.write(", ".join(missing_skills) if missing_skills else "None")
```

```python

%%writefile app.py
import streamlit as st
import numpy as np
from PyPDF2 import PdfReader

st.set_page_config(
    page_title="AI Resume Screening",
    page_icon="📄"
)

st.title("AI-Based Resume Screening System")
st.write("Upload a resume PDF to check its skill-match score.")

required_skills = [
    "Python",
    "SQL",
    "Machine Learning",
    "Pandas",
    "NumPy",
    "Flask"
]

uploaded_file = st.file_uploader(
    "Upload a resume PDF",
    type=["pdf"]
)

if uploaded_file is not None:
    reader = PdfReader(uploaded_file)
    resume_text = ""

    for page in reader.pages:
        text = page.extract_text()
        if text:
            resume_text += text + "\n"

    resume_text_lower = resume_text.lower()

    matched_skills = [
        skill for skill in required_skills
        if skill.lower() in resume_text_lower
    ]

    missing_skills = [
        skill for skill in required_skills
        if skill not in matched_skills
    ]

    score = (
        len(matched_skills) / len(required_skills)
    ) * 100

    st.subheader("Screening Results")
    st.metric("Skill Match Score", f"{score:.2f}%")

    st.write("**Matched skills:**")
    st.write(", ".join(matched_skills) if matched_skills else "None")

    st.write("**Missing skills:**")
    st.write(", ".join(missing_skills) if missing_skills else "None")
```
